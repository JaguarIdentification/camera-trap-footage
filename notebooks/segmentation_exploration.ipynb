# Background influence 

This notebook analyzes how background features influence jaguar re-identification models.

## Load Dataset from HuggingFace

```python
import fiftyone as fo
import fiftyone.zoo as foz
from huggingface_hub import hf_hub_download
import os
import sys
from pathlib import Path
from tqdm.auto import tqdm as std_tqdm
import numpy as np
from PIL import Image as PILImage

repo_root = Path.cwd().parent
sys.path.insert(0, str(repo_root / "src"))

from jaguars.segmentation.utils import (
    get_detection_mask_image,
    get_segmented_bbox_image,
)
```

```python
from fiftyone.utils.huggingface import load_from_hub

# fo.delete_dataset("JaguarCameraTrap/jaguars_camera_trap_0226_fiftyone")

if fo.dataset_exists("JaguarCameraTrap/jaguars_camera_trap_0226_fiftyone"):
    print("Dataset already exists. Loading...")
    dataset = fo.load_dataset("JaguarCameraTrap/jaguars_camera_trap_0226_fiftyone")
else:
    print("Dataset does not exist. Loading from HuggingFace Hub...")
    dataset = load_from_hub("JaguarCameraTrap/jaguars_camera_trap_0226_fiftyone")
dataset.persistent = True

# session = fo.launch_app(dataset)
```

```python
dataset.persistent = True
```

### Explore sample

```python
sample = dataset.first()
print("Filepath:", sample.filepath)

print("Ground Truth:", sample.ground_truth)
print("PPTX Detection?:", sample.pptx_detections)
print("Segmentation exists:", sample.sam3_segmentations)

# Visualize the sample if it has an image
if sample.filepath:
    from PIL import Image
    image = Image.open(sample.filepath)
    display(image)
```

```python
sample = dataset.first()
mask_image = get_detection_mask_image(sample)
if mask_image:
    display(mask_image)
else:
    print("No mask found in detection.")
```

```python
sample = dataset.first()
seg_bbox_image = get_segmented_bbox_image(sample)
if seg_bbox_image:
    display(seg_bbox_image)
```

```python
from jaguars.segmentation.utils import remove_jaguar_from_image

background_only_image = remove_jaguar_from_image(sample)
if background_only_image:
    display(background_only_image)
```

```python
from jaguars.segmentation.utils import canny_edge_detection

edge_image = canny_edge_detection(image=seg_bbox_image)
if edge_image:
    display(edge_image)
```
